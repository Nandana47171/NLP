```python

!pip install nltk

import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
import string
text = """
My name is Nandana Ramachandran. I am from Angamaly, Kerala.
I completed my schooling at M.G.M. Higher Secondary School, Nayathode.
There are four members in my family.
I am currently pursuing B.Tech in Computer Science and Engineering with Artificial Intelligence at Adi Shankara Institute of Engineering and Technology, Kalady.
I have participated in hackathons, completed internships in Python Full Stack Development and Automation Testing, and worked on AI-based projects.
My hobbies are reading, listening to music, and travelling.
I am interested in Artificial Intelligence, Machine Learning, and software development.
My strengths are quick learning, teamwork, and dedication.
My weakness is that I sometimes overthink and take extra time to make decisions.
"""

# Original Text
print("========== ORIGINAL TEXT ==========\n")
print(text)

# Sentence Tokenization
sentences = sent_tokenize(text)
print("\n========== SENTENCE TOKENIZATION ==========\n")
for sentence in sentences:
    print(sentence)

# Word Tokenization
words = word_tokenize(text)
print("\n========== WORD TOKENIZATION ==========\n")
print(words)

# Convert to Lowercase
lower_words = [word.lower() for word in words]
print("\n========== LOWERCASE ==========\n")
print(lower_words)

# Remove Punctuation
no_punctuation = [word for word in lower_words if word not in string.punctuation]
print("\n========== AFTER REMOVING PUNCTUATION ==========\n")
print(no_punctuation)

# Remove Stopwords
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in no_punctuation if word not in stop_words]
print("\n========== AFTER REMOVING STOPWORDS ==========\n")
print(filtered_words)

# Stemming
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_words]
print("\n========== STEMMING ==========\n")
print(stemmed_words)

# Lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]
print("\n========== LEMMATIZATION ==========\n")
print(lemmatized_words)
```
